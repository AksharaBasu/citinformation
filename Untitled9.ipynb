```python
# ============================================================
# CREATE DOCUMENTS AND CHUNKS
# ============================================================

documents = [
    {
        "text": "Witness A says: The suspect was wearing a red hat and left in a blue Honda.",
        "source": "Detective Data"
    },
    {
        "text": "Witness B says: The vault was breached at exactly 11:42 PM using a thermal drill.",
        "source": "Detective Data"
    },
    {
        "text": "The college placement drive for 2026 is scheduled to start on August 15th.",
        "source": "Detective Data"
    }
]

print("✅ Detective data loaded")
print("Documents:", len(documents))


# ============================================================
# ADD VTU WEBSITE
# ============================================================

import requests
from bs4 import BeautifulSoup

vtu_url = "https://vtu.ac.in/en/admissions/"

response = requests.get(
    vtu_url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)

response.raise_for_status()

soup = BeautifulSoup(
    response.text,
    "html.parser"
)

for tag in soup.find_all(
    ["script", "style", "noscript"]
):
    tag.decompose()

vtu_text = soup.get_text(
    separator=" ",
    strip=True
)

documents.append({
    "text": vtu_text,
    "source": "VTU Admissions Website"
})

print("✅ VTU website loaded")
print("Total documents:", len(documents))


# ============================================================
# DOWNLOAD AND READ HACKNOVA PDF
# ============================================================

from pypdf import PdfReader
import urllib.request

pdf_url = (
    "https://www.citcoorg.edu.in/wp-content/uploads/2025/09/"
    "Hacknova1.0-Brochure.pdf"
)

pdf_path = "/content/Hacknova1.0-Brochure.pdf"

request = urllib.request.Request(
    pdf_url,
    headers={"User-Agent": "Mozilla/5.0"}
)

with urllib.request.urlopen(
    request,
    timeout=60
) as response:

    pdf_data = response.read()

with open(pdf_path, "wb") as f:
    f.write(pdf_data)

reader = PdfReader(pdf_path)

pdf_text = ""

for page in reader.pages:

    text = page.extract_text()

    if text:
        pdf_text += "\n" + text

documents.append({
    "text": pdf_text,
    "source": "Hacknova 1.0 Brochure"
})

print("✅ Hacknova PDF loaded")
print("PDF pages:", len(reader.pages))
print("Total documents:", len(documents))


# ============================================================
# CREATE CHUNKS
# ============================================================

CHUNK_SIZE = 500
CHUNK_OVERLAP = 50

chunks = []

for document in documents:

    text = document["text"]
    source = document["source"]

    start = 0

    while start < len(text):

        end = min(
            start + CHUNK_SIZE,
            len(text)
        )

        chunk_text = text[start:end].strip()

        if chunk_text:

            chunks.append({
                "text": chunk_text,
                "source": source
            })

        if end >= len(text):
            break

        start = end - CHUNK_OVERLAP


print("\n✅ CHUNKING COMPLETED")
print("Total chunks:", len(chunks))

print("\nFirst chunk:")
print(chunks[0])
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("✅ Embedding model loaded")
```

```python
chunk_texts = [chunk["text"] for chunk in chunks]

embeddings = embedding_model.encode(
    chunk_texts,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

embeddings = embeddings.astype("float32")

print("✅ Embeddings created")
print("Shape:", embeddings.shape)
```

```python
import faiss

# Number of dimensions in each embedding
dimension = embeddings.shape[1]

# Create FAISS index
index = faiss.IndexFlatIP(dimension)

# Add embeddings to FAISS
index.add(embeddings)

print("✅ FAISS index created successfully")
print("Vectors stored:", index.ntotal)
print("Embedding dimension:", dimension)
```

```python
def search_documents(query, top_k=3):

    # Convert the question into an embedding
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    # Search FAISS
    scores, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for score, idx in zip(scores[0], indices[0]):

        if idx == -1:
            continue

        results.append({
            "text": chunks[idx]["text"],
            "source": chunks[idx]["source"],
            "score": float(score)
        })

    return results

print("✅ Search function created")
```

```python
results = search_documents(
    "What was the suspect wearing?",
    top_k=3
)

for i, result in enumerate(results, start=1):

    print("=" * 60)
    print("RESULT", i)
    print("SOURCE:", result["source"])
    print("SCORE:", round(result["score"], 4))
    print("TEXT:", result["text"])
```

```python
import requests
import urllib3
from bs4 import BeautifulSoup

# Hide the SSL warning caused by verify=False
urllib3.disable_warnings(
    urllib3.exceptions.InsecureRequestWarning
)

vtu_url = "https://admissions.vtu.ac.in/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

print("Loading official VTU Admissions Portal...")

try:
    response = requests.get(
        vtu_url,
        headers=headers,
        timeout=30,
        verify=False
    )

    response.raise_for_status()

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    # Remove unwanted HTML
    for tag in soup.find_all(
        ["script", "style", "noscript"]
    ):
        tag.decompose()

    # Extract text
    raw_text = soup.get_text(
        separator="\n",
        strip=True
    )

    # --------------------------------------------------------
    # Remove obvious spam / junk
    # --------------------------------------------------------

    spam_words = [
        "situs slot",
        "terminal4d",
        "sudirman168",
        "sulebet",
        "slot gacor",
        "casino",
        "judi"
    ]

    clean_lines = []

    for line in raw_text.splitlines():

        line = line.strip()

        if not line:
            continue

        line_lower = line.lower()

        # Skip lines containing obvious spam
        if any(
            spam_word in line_lower
            for spam_word in spam_words
        ):
            continue

        clean_lines.append(line)

    vtu_text = "\n".join(
        clean_lines
    )

    print("✅ VTU website loaded successfully")
    print("Characters extracted:", len(vtu_text))

    print("\nFirst 1000 characters:")
    print(vtu_text[:1000])

except requests.exceptions.SSLError:
    print("❌ SSL problem while accessing VTU.")

except requests.exceptions.RequestException as e:
    print("❌ VTU request failed:")
    print(e)
```
