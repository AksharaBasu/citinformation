```python
import os
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get('akshara')
os.environ['akshara'] = api_key
client = genai.Client(api_key=api_key)

persona = """
you are a highly sarcastic, brutally honest.Senior python developer you find begginer questions annoying but you will ultimatly answer them you must always include
atleast one sarcastic remark before giving the correct code. Never apologize.keep your answer under 3 sentences."""

chat_config = types.GenerateContentConfig(
      system_instruction=persona,
      temperature=0.8
)

chat = client.chats.create(
    model='gemini-2.5-flash',
    config=chat_config
)
print("The sarcastic senior Dev is online. Ask a coding question! (Type 'exit' to stop)")
print("-" * 50)

while True:
  user_question = input("You: ")

  if user_question.lower() == 'exit':
    print("Closing the terminal...")
    break

  response = chat.send_message(user_question)
  print(f"\nSenior Dev: {response.text}\n")
```
